```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

print("Librires imported successfully.")
```

```python
data = pd.DataFrame({
    "Study_Hours": [2, 4, 5, 7, 8, 10],
    "Attendance_Percent": [50, 60, 65, 75, 85, 95],
    "Assignment_Score": [35, 45, 55, 65, 80, 95],
    "Final_Marks": [38, 48, 56, 68, 82, 94]
})

display(data)
```

```python
features = data[["Study_Hours","Attendance_Percent","Assignment_Score"]]

print(features.min())

print(features.max())
```

```python
features.plot(kind="box", figsize=(10, 7))
plt.title("Original Features: Different Scales")
plt.ylabel("Original values")
plt.xlabel("Features")
plt.xticks(rotation=15)
plt.show()
```

```python
normalize = MinMaxScaler()
normalized_array = normalize.fit_transform(features)

normalized_df = pd.DataFrame(normalized_array, columns=features.columns,index=features.index)
display(normalized_df.round(3))
```

```python
print("Maximum after normalization: ")
print(normalized_df.max())

print("Minimum after normalization: ")
print(normalized_df.min())
```

```python
normalized_df.plot(kind="box", figsize=(10, 7))
plt.title("Normalized Features: Same Scales")
plt.ylabel("Normalized values")
plt.xlabel("Features")
plt.xticks(rotation=15)
plt.show()
```

```python
standardizer = StandardScaler()
standardized_array = standardizer.fit_transform(features)

standardized_df = pd.DataFrame(standardized_array, columns=features.columns,index=features.index)
display(standardized_df.round(3))

print("Maximum after normalization: ")
print(standardized_df.max())

print("Minimum after normalization: ")
print(standardized_df.min())

standardized_df.plot(kind="box", figsize=(10, 7))
plt.title("Normalized Features: Same Scales")
plt.ylabel("Normalized values")
plt.xlabel("Features")
plt.xticks(rotation=15)
plt.show()
```

```python
print("Mean after standerdization: ")
print(standardized_df.mean().round(6))

print("Standard deviation after standerdization: ")
print(standardized_df.std().round(6))
```

```python
standardized_df.plot(kind="box", figsize=(10, 7))
plt.title("Features After Standardization")
plt.ylabel("Standardized values")
plt.xlabel("Features")
plt.xticks(rotation=15)
plt.show()
```

```python
scaled_features = StandardScaler().fit_transform(features)
pca = PCA(n_components=2)
pca_array = pca.fit_transform(scaled_features)

pca_df = pd.DataFrame(pca_array, columns=["PC1", "PC2"], index = features.index)
display(pca_df.round(3))
```

```python
variance_ratios = pca.explained_variance_ratio_
for component, ratio in zip(["PC1","PC2"],variance_ratios):
  print(f"{component} explains {ratio:.2%} of the variance")

total_retained = variance_ratios.sum()
print(f"Total variance retained: {total_retained:.2%}")
```

```python
component_names=["PC1","PC2"]
cummulative_variance_ratios=variance_ratios.cumsum()

plt.figure(figsize=(7,4))
plt.bar(component_names,variance_ratios)
plt.plot(component_names,variance_ratios,marker="o")
plt.ylim(0,1.05)
plt.ylabel("Proportion of Variance")
plt.title("PCA Explained and Cummulative variance")
plt.legend(["Cummulative variance","Individual component"])
plt.show()
```

```python
print("Number of original input features:", features.shape[1])
print("Number of principal components:", pca_df.shape)
print("Original feature shape:", features.shape)
print("Reduced PCA data shape:", pca_df.shape)
```
